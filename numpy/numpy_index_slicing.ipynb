```python
import numpy as np

print("version : ",np.__version__)
np_array = [1,2,3,5,8]
print("array : ",np_array)
```

```python
np_array = np.array([
    [1,3],
    [2,4]
])
print("dimensional : ",np_array.ndim)
print("row & column : ",np_array.shape)
print("size : ",np_array.size)
print("data type : ",np_array.dtype)
```

```python
numbers = [1,2,3,4,5]
array = np.array(numbers)
square = array ** 2
print("square : ",square)
```

```python
zeros = np.zeros((3,4) , dtype=int)
print(zeros)
ones = np.ones((3,3), dtype=int)
print(ones)
values = np.arange(1,10,2)
print(values)
```

```python
array = np.array([
    [1,22,63],
    [69,47,27]
])
print(array[-1,-1])
array[0,2] = 99
print(array)
print(array[::-1])
```

```python
array = np.linspace(0,1,6)
print(array)
```

```python
array = np.arange(11)
print(array)
```

```python
identity_matrix = np.eye(3,dtype=int)
print(identity_matrix)
```

```python
random_array = np.random.randint(0,9 , size=(3,3))
print(random_array)
```

```python
array = np.array([10,20,30,40,50,60])
print(array[-2:])
print(array[2:4:2])
print(array[::-1])
```
