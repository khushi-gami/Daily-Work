```python
a = np.array([10,20,30])
b = np.array([40,50,60])

stack_arr = np.stack((a,b))
vstack_arr = np.vstack((a,b))
hstack_arr = np.hstack((a,b))
col_stack = np.column_stack((a,b))

print("stack : \n",stack_arr)
print("vstack : \n",vstack_arr)
print("hstack : \n",hstack_arr)

print("column stack : \n",col_stack)
```

```python
import numpy as np
a = np.array([10, 20, 30, 40, 50, 60])
b = np.array([10, 20, 30, 40, 50])

split_a = np.split(a,3) # equal parts
split_arr = np.array_split(b,3) # unequal parts
print("split : ",split_a)
print("array_split :",split_arr)
```
